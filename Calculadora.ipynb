```python
# Pedimos al usuario que introduzca el primer número y lo convertimos de texto a entero
number1 = int(input("Inserta el numero 1 :")) "Pide al usuario que introduzca un número, conviértelo de texto a número entero y guarda ese número en la variable number1."

# Pedimos al usuario que introduzca el segundo número y lo convertimos de texto a entero
number2 = int(input("Inserta el numero 2 :"))

# Creamos una función llamada sum que recibe los números number1 y number2
def sum(number1, number2):

    # Sumamos number1 y number2 y guardamos el resultado en la variable result
    result = number1 + number2

    # Mostramos por pantalla los dos números, la operación de suma y el resultado
    print(number1, "+", number2, "=", result)

# Creamos una función llamada sub que recibe los números number1 y number2
def sub(number1, number2):

    # Restamos number2 a number1 y guardamos el resultado en la variable result
    result = number1 - number2

    # Mostramos por pantalla los dos números, la operación de resta y el resultado
    print(number1, "-", number2, "=", result)

# Creamos una función llamada mult que recibe los números number1 y number2
def mult(number1, number2):

    # Multiplicamos number1 por number2 y guardamos el resultado en la variable result
    result = number1 * number2

    # Mostramos por pantalla los dos números, la operación de multiplicación y el resultado
    print(number1, "*", number2, "=", result)

# Creamos una función llamada div que recibe los números number1 y number2
def div(number1, number2):

    # Comprobamos si el primer número es diferente de cero
    if (number2 != 0):

        # Dividimos number1 entre number2 y guardamos el resultado en la variable result
        result = number1 / number2

        # Mostramos por pantalla los dos números, la operación de división y el resultado
        print(number1, "/", number2, "=", result)

    # Si number1 es igual a cero, ejecutamos las instrucciones del bloque else
    else:

        # Mostramos un mensaje indicando que el número 2 es cero y no se puede dividir entre cero
        print("El numero 2 es igual a 0 y no se puede dividir por 0")

# Llamamos a la función sum y le pasamos los dos números introducidos por el usuario
sum(number1, number2)

# Llamamos a la función sub y le pasamos los dos números introducidos por el usuario
sub(number1, number2)

# Llamamos a la función mult y le pasamos los dos números introducidos por el usuario
mult(number1, number2)

# Llamamos a la función div y le pasamos los dos números introducidos por el usuario
div(number1, number2)
```
